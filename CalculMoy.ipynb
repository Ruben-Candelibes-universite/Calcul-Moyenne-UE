# **Calcul des Moyennes des UE**

<u> **Version 1 :** </u>

<u> Aglorithme </u>

demande les notes de chaque Ressources et de chaque coeff de cette UE
faire la moyenne de la note avec le coeff
add les moyennes entre elle puis diviser le total pour avoir la moyenne de l'UE

```python
def Moy():
    
    moy = 0
    total_coeff = 0
    note = 0
    coeff = 0

    for r in range(101,117):
        note = float(input("Indiquez votre note de la ressource R{} : ".format(r)))
        coeff = float(input("Indiquez le coeff de la ressource R{} : ".format(r)))
        moy = moy + note * coeff
        total_coeff = total_coeff + coeff

    for s in range(11,17):
        note = float(input("Indiquez votre note de la SAE{} : ".format(s)))
        coeff = float(input("Indiquez le coeff de la SAE{} : ".format(s)))
        moy = moy + note * coeff
        total_coeff = total_coeff + coeff

    moy = moy / total_coeff
    print("Voici la moyenne de l'UE :", moy)


Moy()
```
