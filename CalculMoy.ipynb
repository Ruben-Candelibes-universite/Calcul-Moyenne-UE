<u> **Version 1 :** </u>

<u> Aglorithme </u>

demande les notes de chaque Ressources et de chaque coeff de cette UE
faire la moyenne de la note avec le coeff
add les moyennes entre elle puis diviser le total pour avoir la moyenne de l'UE

```python
# Version 1 du calcul des moyennes des UE

def Moy():
    
    moy = 0              # Contient la somme des notes multipliées par leur coefficient
    total_coeff = 0      # Contient la somme de tous les coefficients
    note = 0             # Stocke temporairement la note saisie
    coeff = 0            # Stocke temporairement le coefficient saisi

    for r in range(101,103):     # Boucle pour les ressources R101 à R116

        note = float(input("Indiquez votre note de la ressource R{} : ".format(r)))       # Demande la note de la ressource        
        coeff = float(input("Indiquez le coeff de la ressource R{} : ".format(r)))        # Demande le coefficient de la ressource
        moy = moy + note * coeff                                                          # Ajoute note × coefficient à la somme
        total_coeff = total_coeff + coeff                                                 # Ajoute le coefficient au total des coefficients

    for s in range(11,14):       # Boucle pour les SAE11 à SAE16 
        
        note = float(input("Indiquez votre note de la SAE{} : ".format(s)))               # Demande la note de la SAE
        coeff = float(input("Indiquez le coeff de la SAE{} : ".format(s)))                # Demande le coefficient de la SAE     
        moy = moy + note * coeff                                                          # Ajoute note × coefficient à la somme        
        total_coeff = total_coeff + coeff                                                 # Ajoute le coefficient au total des coefficients
    
                    
    moy = moy / total_coeff                      # Calcule la moyenne 
    print("Voici la moyenne de l'UE :", moy)     # Affiche la moyenne finale



Moy()
```

```python
import matplotlib.pyplot as plt

matiere = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)] # Crée une liste des matière pour les ressources et les SAE
note = [float(input(f"Quelle est ta note pour {elt} ?")) for elt in matiere] # Demande la note à l'utilisateur de chaque matière, la note est transformé en nombre décimal (float)
moyennes = [moy1, moy2, moy3] # Les moyennes calculées dans le bloc précédent

# Liste des coefficients pour chaque matière
coef1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
coef2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
coef3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

# Crée un dictionnaire associant chaque matière à sa note
dicnote = {cle: valeur for cle, valeur in zip(matiere, note)}  
dicoef1 = {cle: valeur for cle, valeur in zip(matiere, coef1)} 
dicoef2 = {cle: valeur for cle, valeur in zip(matiere, coef2)}  
dicoef3 = {cle: valeur for cle, valeur in zip(matiere, coef3)}  

# Multiplie la note de chaque matière par son coef
dicoresult1={matiere:dicnote[matiere]*dicoef1[matiere] for matiere in dicnote} 
dicoresult2={matiere:dicnote[matiere]*dicoef2[matiere] for matiere in dicnote} 
dicoresult3={matiere:dicnote[matiere]*dicoef3[matiere] for matiere in dicnote}

# calcule la moyenne le sum permet d'additionner toutes les notes et coef 
moy1=sum(dicoresult1.values())/sum(dicoef1.values())
moy2=sum(dicoresult2.values())/sum(dicoef2.values()) 
moy3=sum(dicoresult3.values())/sum(dicoef3.values())

for moyenne in moyennes:
    if moyenne < 8:
        print("Vous ne passez pas")
    elif moyenne < 10:
        print("C'est bon mais attention")
    else:
        print("Nickel c'est bon pour la suite !")

# Affiche la moyenne obtenue 
print(moy1)
print(moy2)
print(moy3)


ue = ["UE 1", "UE 2", "UE 3"] # Inisialise les 3 UE pour l'axe x
couleurs = [] # Inisialise les couleur de chaque barre

for moyenne in moyennes: # On regarde si moyennes est en dessous de 8 ou non 
    if moyenne>=10:
        couleurs.append("green")
    elif 8<=moyenne<10:
        couleurs.append("orange")
    else:
        couleurs.append("red")
plt.close() # Ferme les autres graphiques 
plt.bar(ue, moyennes, color=couleurs) # Création du graphique
plt.ylim(0, 20) # Limites de l'axe Y
# Nom des axes
plt.xlabel("UE du Semestre 1") 
plt.ylabel("Moyenne sur 20")
# Titre
plt.title("Moyennes des UE") # Titre
# Affiche la valeur au-dessus de chaque barre

for i, moyenne in enumerate(moyennes):
    plt.text(i, moyenne + 0.3, f"{moyenne:.2f}", ha="center")

plt.show() # Affiche le graphique
```
